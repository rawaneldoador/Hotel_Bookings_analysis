```python
#Dealing with missing values
```

```python
missings = df.isna().mean()*100
print(missings[missings > 0].sort_values(ascending=False))
```

```python
df.drop('company', axis=1, inplace=True)
```

```python
df['agent'] = df['agent'].fillna(0)
```

```python
df['country']=df['country'].fillna(df['country'].mode()[0])
```

```python
df['children']=df['children'].fillna(0)
```

```python
#Data Type Conversion
```

```python
df['children'] = df['children'].astype(int)
```

```python
df['agent'] = df['agent'].astype(int)
```

```python
df['reservation_status_date'] = pd.to_datetime(df['reservation_status_date'], errors='coerce')
df['reservation_status_date'] = df['reservation_status_date'].dt.strftime('%Y-%m-%d')
```

```python
for col in ['hotel', 'meal', 'country', 'market_segment', 'distribution_channel', 'reserved_room_type', 'assigned_room_type', 'deposit_type', 'customer_type', 'reservation_status']:
    df[col] = df[col].astype('category')
```

```python
numeric = df.select_dtypes(include=['number']).columns  
negatives = (df[numeric] < 0).any()

print(negatives)
```

```python
Q1 = df['adr'].quantile(0.25)
Q3 = df['adr'].quantile(0.75)
IQR = Q3 - Q1 
lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR
 
df= df[(df['adr'] >= lower_bound) & (df['adr'] <= upper_bound)]
sns.boxplot(df['adr'])
```

```python
print("Booking Cancellation Status:\n")
cancel_counts = df['is_canceled'].value_counts()
cancel_percentages = df['is_canceled'].value_counts(normalize=True) * 100

print(f"Not Canceled: {cancel_counts[0]} bookings ({cancel_percentages[0]:.2f}%)")
print(f"Canceled: {cancel_counts[1]} bookings ({cancel_percentages[1]:.2f}%)")
```

```python
print("Monthly Bookings (Sorted Descending by Count):\n")

monthly_counts = df['arrival_date_month'].value_counts()
monthly_percentages = df['arrival_date_month'].value_counts(normalize=True) * 100

sorted_months = monthly_counts.sort_values(ascending=False).index

for month in sorted_months:
    count = monthly_counts[month]
    percentage = monthly_percentages[month]
    print(f"{month}: {count} bookings ({percentage:.2f}%)")
```

```python
print("Number of Bookings per Year (Sorted Descending):\n")

year_counts = df['arrival_date_year'].value_counts().sort_values(ascending=False)
year_percentages = df['arrival_date_year'].value_counts(normalize=True).sort_values(ascending=False) * 100

for year in year_counts.index:
    count = year_counts[year]
    percentage = year_percentages[year]
    print(f"{year}: {count} bookings ({percentage:.2f}%)")
```

```python
print("Top 10 Countries for City Hotel Bookings:\n")

top_countries_city = df[df['hotel'] == 'City Hotel']['country'].value_counts().head(10)

for country, count in top_countries_city.items():
    print(f"{country}: {count} bookings")

print("\nTop 10 Countries for Resort Hotel Bookings:\n")

top_countries_resort = df[df['hotel'] == 'Resort Hotel']['country'].value_counts().head(10)

for country, count in top_countries_resort.items():
    print(f"{country}: {count} bookings")
```

```python
print("Top 10 Countries by Number of Bookings per Hotel Type:\n")
country_hotel = df.groupby(['country', 'hotel']).size().unstack().fillna(0)


top_10 = df['country'].value_counts().head(10).index


country_hotel_top = country_hotel.loc[top_10].astype(int)


for country in country_hotel_top.index:
    print(f"{country}:")
    for hotel_type in country_hotel_top.columns:
        count = country_hotel_top.loc[country, hotel_type]
        print(f"  {hotel_type}: {count} bookings")
    print()
```

```python
print("Customer Type Distribution (Sorted Descending):\n")

customer_counts = df['customer_type'].value_counts().sort_values(ascending=False)
customer_percentages = df['customer_type'].value_counts(normalize=True).sort_values(ascending=False) * 100

for customer_type in customer_counts.index:
    count = customer_counts[customer_type]
    percentage = customer_percentages[customer_type]
    print(f"{customer_type}: {count} bookings ({percentage:.2f}%)")
```

```python
top_countries = df['country'].value_counts().head(10)
print("📌 Top 10 Countries with Most Bookings:")
print(top_countries)
```

```python
import calendar


df['arrival_month_num'] = df['arrival_date_month'].apply(lambda x: list(calendar.month_name).index(x))


def get_season(month):
    if month in [12, 1, 2]:
        return 'Winter'
    elif month in [3, 4, 5]:
        return 'Spring'
    elif month in [6, 7, 8]:
        return 'Summer'
    else:
        return 'Fall'

df['season'] = df['arrival_month_num'].apply(get_season)
```

```python
season_sales = df.groupby('season')['adr'].mean().sort_values(ascending=False)

print("📊 Average Sales (ADR) per Season:")
print(season_sales)
```

```python
ax=sns.countplot(data=df, x='arrival_date_year', hue='hotel',palette='Blues')
total = len(df)
for container in ax.containers:
    for bar in container:
        height = bar.get_height()
        if height > 0:
            percentage = f'{height / total * 100:.1f}%'
            ax.text(
                bar.get_x() + bar.get_width() / 2,  
                height + 1,  
                percentage,
                ha='center',
                va='bottom',
                fontsize=9
            )

plt.title('Years arrivals')
plt.xlabel('Year')
plt.ylabel('Count')
plt.show()
```
